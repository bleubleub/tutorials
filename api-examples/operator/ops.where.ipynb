## mindspore.ops.where(condition, input, other) -〉 Tensor
根据条件从输入中选择元素组成返回tensor。
- 输入：
    * condition：Tensor[bool]。
    * input：Tensor, Scalar。
    * other：Tensor, Scalar。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| condition   | condition   | condition   |
| input       | input       | x           |
| other       | other       | y           |
|             | out         | size        |
|             |             | fill_value  |

* torch额外提供了out出参方式。
* jax提供了size和fill_value用于处理当x和y为空的情况。ms与torch的input和other均为必选参数。

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

input = np.array([[-0.4620,  0.3139],
                  [ 0.3898, -0.7197],
                  [ 0.0478, -0.1657]])

y1 = ms.ops.where(ms.tensor(input) > 0, ms.tensor(input), 0)
y2 = torch.where(torch.tensor(input) > 0, torch.tensor(input), 0)
y3 = jnp.where(input > 0, input, 0)
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

* ms与jax不返回类型。
