## mindspore.ops.meshgrid(*inputs, indexing='xy') -〉 Tensor
从给定的Tensor生成网格矩阵。
- 输入：
    * inputs：tuple[Tensor], list[Tensor], Tensor为mindspore的tensor。
    * indexing：str枚举值：'xy' 或 'ij'。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| inputs      | tensors     | xi          |
| indexing    | indexing    | copy(不可修改)|
|             |             | sparse      |
|             |             | indexing    |

* jax额外提供了sparse参数用于将每维度的元素降为1。

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

x = np.array([1, 2, 3, 4])
y = np.array([5, 6, 7])
z = np.array([8, 9, 0, 1, 2])

y1 = ms.ops.meshgrid(ms.tensor(x), ms.tensor(y), ms.tensor(z), indexing='xy')
y2 = torch.meshgrid(torch.tensor(x), torch.tensor(y), torch.tensor(z), indexing='xy')
y3 = jnp.meshgrid(x, y, z, indexing='xy')
print ('mindspore output:\n',y1)
print ('torch output:\n',y2)
print ('jax output:\n',y3)
```

```python
y1 = ms.ops.meshgrid(ms.tensor(x), ms.tensor(y), ms.tensor(z), indexing='ij')
y2 = torch.meshgrid(torch.tensor(x), torch.tensor(y), torch.tensor(z), indexing='ij')
y3 = jnp.meshgrid(x, y, z, indexing='ij')
print ('mindspore output:\n',y1)
print ('torch output:\n',y2)
print ('jax output:\n',y3)
```

* ms返回值的value在最后面，建议关键信息提前。
