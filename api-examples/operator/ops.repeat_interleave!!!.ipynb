## mindspore.ops.repeat_interleave(input, repeats, axis=None) -〉 Tensor
按指定轴重复Tensor的元素。
- 输入：
    * input：mindspore的tensor。
    * repeats：int, tuple, list, mindspore的tensor。
    * axis：int。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| input       | input       | a           |
| repeats     | repeats     | repeats     |
| axis        | dim         | axis        |
|             | output_size | total_repeat_length |

* torch与jax提供了output_size和total_repeat_length参数，用于设定重复次数。

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

input = np.array([[1, 2],
                  [3, 4]])

y1 = ms.ops.repeat_interleave(ms.tensor(input), 2, 1)
y2 = torch.repeat_interleave(torch.tensor(input), 2, 1)
y3 = jnp.repeat(input, 2, 1)
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

```python
y1 = ms.ops.repeat_interleave(input, 2, 1)
print ('mindspore output:\n',y1)

y1 = ms.ops.repeat_interleave(torch.tensor(input), 2, 1)
print ('mindspore output:\n',y1)
```

* 实测发现ms可接受array/torch.tensor输入，文档未体现, 且与其他接口不一致。
* ms与jax不返回类型。

设置重复次数：

```python
# y2 = torch.repeat_interleave(torch.tensor(input), 2, 1, output_size=6)
y3 = jnp.repeat(input, 2, 1, total_repeat_length=6)
# print ('torch output:\n',y2)
print ('jax output:\n',y3)
```

除4外torch不能设置其他值，jax可随意设置。
