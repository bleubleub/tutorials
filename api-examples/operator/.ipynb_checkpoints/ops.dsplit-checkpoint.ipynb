## mindspore.ops.dsplit(input, indices_or_sections) -〉 tuple[Tensor]
沿着第三轴将输入Tensor分割成多个子Tensor。等同于axis = 2时的 ops.tensor_split 。
- 输入：
    * input (Tensor)
    * indices_or_sections (Union[int, tuple(int), list(int)])
- 返回：mindspore的tensor。

## 1、参数比较：
| mindspore   | torch       |
| :----:      | :----:      |
| input       | input       |
| indices_or_sections  | indices_or_sections        |

## 2、返回值比较  
当indices_or_sections 为 int时：

```python
import numpy as np
import torch
import mindspore as ms
from mindspore import Tensor, ops

input = np.arange(16.0).reshape(2, 2, 4)
print("input:\n", input)
print("\n")
output1 = torch.dsplit(torch.tensor(input), 2)
print("torch output:\n", output1)
print("\n")
output2 = ms.ops.dsplit(ms.tensor(input), 2)
print("ms output:\n", output2)
```

当indices_or_sections 为list时：

```python
output1 = torch.dsplit(torch.tensor(input), [3, 6])
print("torch output1:\n", output1)
print("\n")
output2 = ms.ops.dsplit(ms.tensor(input), [3, 6])
print("ms output2:\n", output1)
```

### 与tensor_split接口比较：

```python
output1 = torch.tensor_split(torch.tensor(input), 2, 2)
print("torch output1:\n", output1)
output2 = torch.tensor_split(torch.tensor(input), [3, 6], 2)
print("torch output2:\n", output2)
```

```python
output1 = ms.ops.tensor_split(ms.tensor(input), 2, 2)
print("ms output1:\n", output1)
output2 = ms.ops.tensor_split(ms.tensor(input), [3, 6], 2)
print("ms output2:\n", output2)
```

```python
a = torch.arange(10).reshape(5, 2)
torch.split(a, 3)
```

当axis=2时，ms与torch的dsplit与tensor_split接口返回值一致。
但ms的两个接口输出格式有区别。

所以为了得到同样的输出，入参需要改变，如下：

```python
output1 = torch.split(torch.tensor(input), 2, 2)
print("torch output1:\n", output1)
output2 = torch.split(torch.tensor(input), [3, 1], 2)
print("torch output2:\n", output2)
```
