## mindspore.ops.reciprocal(input) -〉 Tensor
返回输入的每个元素的倒数。
- 输入：input必须为mindspore的tensor。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| input       | input       | x           |
|             | out         |             |

* torch额外提供了out出参方式

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

input = np.array([1.0, 2.0, 4.0])

y1 = ms.ops.reciprocal(ms.tensor(input))
y2 = torch.reciprocal(torch.tensor(input))
y3 = jnp.reciprocal(input)
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

* ms与jax不返回类型。
