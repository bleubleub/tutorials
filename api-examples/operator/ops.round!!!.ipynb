## mindspore.ops.round(input, *, decimals=0) -〉 Tensor
返回输入四舍五入后的整数值。
- 输入：
    * input：mindspore的tensor。
    * decimals：int, 要保留的小数位数。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| input       | input       | a           |
| decimals    | decimals    | decimals    |
|             | out         | out(unused) |

* torch额外提供了out出参方式。

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

input = np.array([1.532, 3.267, 6.149])

y1 = ms.ops.round(ms.tensor(input), decimals=2)
print ('mindspore output:\n',y1)
```

```python
y2 = torch.round(torch.tensor(input), decimals=2)
y3 = jnp.round(input, 2)
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

* 当设置decimals=2时，报错只支持decimals为0，不太合理。jax与torch均能配置。
* ms与jax不返回类型。

```python
input = np.array([10.5, 21.5, 12.5, 31.5])

y1 = ms.ops.round(ms.tensor(input))
y2 = torch.round(torch.tensor(input))
y3 = jnp.round(input)
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

面对0.5的情况，三种框架表现一致。
