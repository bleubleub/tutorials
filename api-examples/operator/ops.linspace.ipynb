## mindspore.ops.linspace(start, end, steps) -〉 Tensor
返回一个在区间start和end内的包含steps个值的一维Tensor。
- 输入：
    * start：int、float或mindspore的tensor（数据类型为float32）。
    * end：int、float或mindspore的tensor（数据类型为float32）。
    * steps：int或mindspore的tensor，必须为正数。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| start       | start       | start       |
| end         | end         | stop        |
| steps       | steps       | num         |
|             | out         | endpoint    |
|             | dtype       | retstep     |
|             | layout      | dtype       |
|             | device      | axis        |
|             | requires_grad| device     |

* torch和jax都提供了device参数，用于设置返回值提交到哪个设备上。
* torch额外提供了out出参，layout用于设置内存存储结构，以及requires_grad设置自动求导是否记录操作。
* jax额外提供了endpoint、retstep、axis参数用于更细节的调整返回的数组。

2、返回值比较

生成一维数组：

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

y1 = ms.ops.linspace(0, 10, 5)
y2 = torch.linspace(0, 10, 5)
y3 = jnp.linspace(0, 10, 5)
print ('mindspore output:',y1)
print ('torch output:',y2)
print ('jax output:',y3)
```

生成多维数组：

```python
y3 = jnp.linspace(np.array([0, 5]), np.array([5, 10]), 5)
print ('jax output:',y3)
```

* ms和torch的前两个参数都不支持传入多维tensor，所以无法生成多维tensor。jax支持。

3、报错信息比较

```python
y1 = ms.ops.linspace([0], [10], [5])
```

```python
y2 = torch.linspace([0], [10], [5])
```

```python
y3 = jnp.linspace([0], [10], [5])
```

当输入类型不正确时，建议ms优化报错信息。
