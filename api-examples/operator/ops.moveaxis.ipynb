## mindspore.ops.moveaxis(x, source, destination) -〉 Tensor
调换 x 中 source 和 destination 两个维度的位置。
- 输入：
    * x：mindspore的tensor。
    * source：int, sequence[int]。
    * destination：int, sequence[int]。
- 返回：mindspore的tensor。

1、参数比较：

| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| x           | input       | a           |
| source      | source      | source      |
| destination | destination | destination |


2、返回值比较

实数计算：

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

input = np.ones((2, 3, 4, 5))

y1 = ms.ops.moveaxis(ms.tensor(input), 1, -1)
y2 = torch.moveaxis(torch.tensor(input),1,-1)
y3 = jnp.moveaxis(input, 1, -1)
print ('mindspore output:\n',y1.shape)
print('\n')
print ('torch output:\n',y2.shape)
print('\n')
print ('jax output:\n',y3.shape)
```

复数计算:

```python
y1 = ms.ops.moveaxis(ms.tensor(input), (0, 1), (-1, -2))
y2 = torch.moveaxis(torch.tensor(input),(0, 1), (-1, -2))
y3 = jnp.moveaxis(input, (0, 1), (-1, -2))
print ('mindspore output:\n',y1.shape)
print('\n')
print ('torch output:\n',y2.shape)
print('\n')
print ('jax output:\n',y3.shape)
```

* ms与jax不返回类型。
