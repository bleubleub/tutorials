## mindspore.ops.sigmoid(input) -〉 Tensor
逐元素计算sigmoid激活函数。
- 输入：input：mindspore的tensor。
- 返回：mindspore的tensor, 数据类型和shape与input相同。

1、参数比较：
| mindspore   | torch       |
| :----:      | :----:      |
| input       | input       |
|             | out         |

2、返回值比较

```python
import mindspore as ms
import torch
import jax.numpy as jnp

input = [ 5,  1.0887, -0.8858, -1.7683]

y1 = ms.ops.sigmoid(ms.tensor(input))
y2 = torch.sigmoid(torch.tensor(input))
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
```

* ms不返回类型。

3、报错信息比较

```python
y1 = ms.ops.sigmoid(input)
```

```python
y2 = torch.sigmoid(input)
```

当输入类型不正确时，报错信息torch简洁明确。建议ms优化。
