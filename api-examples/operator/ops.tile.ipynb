## mindspore.ops.tile(input, dims) -〉 Tensor
根据指定维度复制input。
- 输入：
    * input：mindspore的tensor。
    * dims：tuple[int]。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| input       | input       | a           |
| dims        | dims        | reps        |

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

input = np.array([[1, 2],
                  [3, 4,]])

y1 = ms.ops.tile(ms.tensor(input), (2, 3))
y2 = torch.tile(torch.tensor(input), (2, 3))
y3 = jnp.tile(input, (2, 3))
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

* ms与jax不返回类型。
