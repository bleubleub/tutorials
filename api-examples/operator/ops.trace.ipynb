## mindspore.ops.trace(input) -〉 Tensor
返回二维输入在主对角线方向上的元素总和。
- 输入：input必须是mindspore的二维tensor, 数据类型不支持复数。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| input       | input       | a           |
|             |             | offset      |
|             |             | axis1       |
|             |             | axis2       |
|             |             | dtype       |
|             |             | out (unused)|

* jax额外提供了offest、axis1、axis2、dtype参数，分别用于设置对角线偏移量、沿哪个维度进行求和以及返回值的数据类型。

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

input = np.array([[1, 2],
                  [3, 4],
                  [5, 6],
                  [7, 8]])

y1 = ms.ops.trace(ms.tensor(input))
y2 = torch.trace(torch.tensor(input))
y3 = jnp.trace(input)
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

* ms与jax不返回类型。

jax使用offest、axis1、axis2参数：

```python
input = np.array([[[1, 2],
                   [3, 4]],

                  [[5, 6],
                   [7, 8]]])

y3 = jnp.trace(input)
print ('jax output:\n',y3)
```

```python
y3 = jnp.trace(input, offset = 1)
print ('jax output:\n',y3)
```

```python
y3 = jnp.trace(input, axis1=1, axis2=2)
print ('jax output:\n',y3)
```

* ms与torch仅能计算二维数组，jax可以计算多维数组。
