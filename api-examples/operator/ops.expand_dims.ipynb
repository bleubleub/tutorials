## mindspore.ops.expand_dims(input_x, axis) -〉 Tensor
对输入在给定的轴上添加额外维度。
- 输入：
    * input_x：mindspore的tensor。
    * axis：int。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   |jax         |
| :----:      |:----:      |
| input_x     |a           |
| axis        |axis        |

2、返回值比较

```python
import numpy as np
import mindspore as ms
import jax.numpy as jnp

input = np.array([1, 2, 3])

y1 = ms.ops.expand_dims(ms.tensor(input), 0)
y2 = jnp.expand_dims(input, 0)
print ('mindspore output:\n',y1, y1.shape)
print('\n')
print ('jax output:\n',y1, y2.shape)
```

```python
y1 = ms.ops.expand_dims(ms.tensor(input), 1)
y2 = jnp.expand_dims(input, 1)
print ('mindspore output:\n',y1, y1.shape)
print('\n')
print ('jax output:\n',y1, y2.shape)
```

```python
y2 = jnp.expand_dims(input, (0, 1, 3))
print ('jax output:\n',y1, y2.shape)
```

* jax的axis参数可以接受传入tuple, ms不接受，仅能传入int。

3、报错信息比较
