## mindspore.ops.tensor_dot(x1, x2, axes) -〉 Tensor
在指定轴上对两个输入进行点乘操作。
- 输入：
    * x1：mindspore的tensor, 数据类型为float16或float32。
    * x2：mindspore的tensor, 数据类型为float16或float32。
    * axes：int, tuple(int), tuple(tuple(int)), list(list(int))。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| x1          | a           | a           |
| x2          | b           | b           |
| axes        | dims        | axes        |
|             | out         | precision   |
|             |             | preferred_element_type |
* torch额外提供了out出参
* jax额外提供了precision用于设置后端精度，以及preferred_element_type设置返回值的数据类型。

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

x1 = np.arange(24.).reshape(2, 3, 4)
x2 = np.ones((3, 4, 5))

y1 = ms.ops.tensor_dot(ms.tensor(x1), ms.tensor(x2), axes=([1, 2], [0, 1]))
y2 = torch.tensordot(torch.tensor(x1), torch.tensor(x2), dims=([1, 2], [0, 1]))
y3 = jnp.tensordot(x1, x2, axes=([1, 2], [0, 1]))
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

* ms与jax不返回类型。
