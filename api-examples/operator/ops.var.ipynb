## mindspore.ops.var(input, axis=None, ddof=0, keepdims=False) -〉 Tensor
按指定维度返回输入Tensor上的方差。
- 输入：
    * input：mindspore的tensor(数据类型为Number)。
    * axis：int, tuple(int)。
    * ddof：int, bool。
    * keepdims：bool。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| input       | input       | a           |
| axis        | dim         | axis        |
| ddof        | correction  | dtype       |
| keepdims    | keepdim     | out (unused)|
|             | out         | ddof        |
|             |             | keepdims    |
|             |             | where       |
|             |             | correction  |

* jax额外提供了dtype、where和correction参数，分别用于设置返回值的数据类型、数组位置上的元素是否参与计算方差。
* torch额外提供了out出参。

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

input = np.array([[1., 3, 4, 2],
                  [5, 2, 6, 3],
                  [8, 4, 2, 9]])

y1 = ms.ops.var(ms.tensor(input))
y2 = torch.var(torch.tensor(input))
y3 = jnp.var(input)
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

```python
y1 = ms.ops.var(ms.tensor(input), 1, keepdims=True, ddof=1)
y2 = torch.var(torch.tensor(input), 1, correction=1, keepdim=True)
y3 = jnp.var(input, 1, keepdims=True, ddof=1)
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

* ms与jax不返回类型。
