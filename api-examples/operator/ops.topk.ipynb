## mindspore.ops.topk(input, k, dim=None, largest=True, sorted=True) -〉Tuple(Tensor)
沿给定维度查找 k 个最大或最小元素和对应的索引。
- 输入：
    * input: mindspore的tensor。
    * k: int。
    * dim: int。
    * largest: bool。
    * sorted: bool。
- 返回：由两个mindspore的tensor组成的tuple。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| input       | input       | operand     |
| k           | k           | k           |
| dim         | dim         |             |
| largest     | largest     |             |
| sorted      | sorted      |             |
|             | out         |             |

* ms与torch基本对齐，torch额外提供了out的出参方式。
* jax没有提供dim、largest与sorted参数。不能按维度计算、查找最小值及规定返回值的排序。

2、返回值比较

一维数组：

```python
import numpy as np
import mindspore as ms
import torch
import jax.lax as jlx

input = np.array([ 1.,  2.,  3.,  4.,  5.])

y1 = ms.ops.topk(ms.tensor(input), 3)
y2 = torch.topk(torch.tensor(input), 3)
y3 = jlx.top_k(input, 3)
print ('mindspore output:',y1)
print ('torch output:',y2)
print ('jax output:',y3)
```

二维数组：

```python
input = np.array([[0.5368, 0.2447, 0.4302, 0.9673],
                  [0.4388, 0.6525, 0.4685, 0.1868],
                  [0.3563, 0.5152, 0.9675, 0.8230]])

y1 = ms.ops.topk(ms.tensor(input), 2, dim=0)
y2 = torch.topk(torch.tensor(input), 2, dim=0)
print ('mindspore output:\n',y1)
print ('\n')
print ('torch output:\n',y2)
```

* m与jax都不返回类型
