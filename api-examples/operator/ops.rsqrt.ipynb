## mindspore.ops.rsqrt(input) -〉 Tensor
计算输入的平方根倒数。
- 输入：mindspore的tensor。
- 返回：mindspore的tensor，与input的shape相同。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| x           | input       | x           |
|             | out         |             |

* torch额外提供了out出参方式

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.lax as jlx

input = np.array([-0.0370,  0.2970,  1.5420, -0.9105])

y1 = ms.ops.rsqrt(ms.tensor(input))
y2 = torch.rsqrt(torch.tensor(input))
y3 = jlx.rsqrt(input)
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

* ms与jax不返回类型。
