## mindspore.ops.remainder(input, other) -〉 Tensor
计算第一个输入元素除以第二个输入元素的余数。
- 输入：input和other必须为numbers.Number、bool或mindspore的tensor（数据类型为Number）。
- 返回：mindspore的tensor。

1、参数比较：
| mindspore   | torch       | jax         |
| :----:      | :----:      | :----:      |
| input       | input       | x1          |
| other       | other       | x2          |
|             | out         |             |

* torch额外提供了out出参方式

2、返回值比较

```python
import numpy as np
import mindspore as ms
import torch
import jax.numpy as jnp

input = np.array([[3, -1, 4],
                  [8, 5, -2]])
other = np.array([2, 3, -5])

y1 = ms.ops.remainder(ms.tensor(input), ms.tensor(other))
y2 = torch.remainder(torch.tensor(input), torch.tensor(other))
y3 = jnp.remainder(input, other)
print ('mindspore output:\n',y1)
print('\n')
print ('torch output:\n',y2)
print('\n')
print ('jax output:\n',y3)
```

* ms与jax不返回类型。

3、报错信息比较

```python
y1 = ms.ops.remainder(input, other)
```
